```python
import pickle
import numpy as np
import torch
from torch import nn
from matplotlib import pyplot as plt
from datetime import datetime, timezone
from scipy import ndimage

from Modules import GetRootPath, ResponseFunctionsLoader, RadMonitorDataLoader, FluxParametersToCounts, Dataset
```

```python
# Class definition for pickled object
class NeuralNetwork(nn.Module):
    def __init__(self, inputDim, outputDim, layerWidths):
        super(NeuralNetwork, self).__init__()

        self.inputLayer = nn.Sequential(
            nn.Linear(inputDim, layerWidths[0]),
            nn.ReLU()
        )
        # List comprehension, followed by star expression unpacking
        self.hiddenLayers = nn.Sequential(*[
            nn.Sequential(
                nn.Linear(layerWidths[i], layerWidths[i+1]),
                nn.ReLU()
            ) for i in range(len(layerWidths)-1)
        ])
        self.outputLayer = nn.Linear(layerWidths[-1], outputDim)

    def forward(self, x) -> torch.Tensor:
        x = self.inputLayer(x)
        x = self.hiddenLayers(x)
        x = self.outputLayer(x)
        return x
```

```python
keys = ["PROTONS"]
nChannels = 6
timeStep = 10 # Minutes
listSEPDates = [
    [datetime(2024, 1, 29), datetime(2024, 1, 31)],
    [datetime(2024, 2, 9), datetime(2024, 2, 16)],
    [datetime(2024, 3, 15), datetime(2024, 3, 19)],
    [datetime(2024, 3, 23), datetime(2024, 3, 27)],
    [datetime(2024, 5, 10), datetime(2024, 5, 18)],
    [datetime(2024, 6, 7), datetime(2024, 6, 14)]
]

listSEPData = []
for i in range(len(listSEPDates)):
    dataSEPLoader = RadMonitorDataLoader()
    dataSEPLoader.LoadCDFFiles(keys, *listSEPDates[i])
    listSEPData.append(dataSEPLoader.dataFrames["PROTONS"])

for i in range(len(listSEPData)):
    # Only keeping count rate data from the first five Proton Detector Head channels
    listSEPData[i] = listSEPData[i].iloc[:, :nChannels]
    listSEPData[i] = listSEPData[i].resample(f"{timeStep}min").mean()
    # From counts/min to counts/sec
    listSEPData[i] = listSEPData[i].div(60)
    # Removing electrical noise spikes in PDH2
    listSEPData[i]["PROTONS_2"] = ndimage.median_filter(listSEPData[i]["PROTONS_2"], size=3)

# Plot of the RADEM count rate data
for dataSEP in listSEPData:
    for i, column in enumerate(dataSEP.columns):
        plt.plot(dataSEP.index, dataSEP[column], label=f"PDH{i+1}")
        plt.xticks(rotation=45)
        plt.yscale("log")
    plt.ylabel(r"Count Rate [s$^{-1}$]")
    plt.legend()
    plt.show()

dataBkgLoader = RadMonitorDataLoader()
dataBkgLoader.LoadCDFFiles(keys, datetime(2023, 10, 1), datetime(2023, 11, 1))
dataBkg = dataBkgLoader.dataFrames["PROTONS"]
dataBkg = dataBkg.iloc[:, :nChannels]
dataBkg = dataBkg.resample(f"{timeStep}min").mean()
dataBkg = dataBkg.div(60)
dataBkg["PROTONS_2"] = ndimage.median_filter(dataBkg["PROTONS_2"], size=3)

for i, column in enumerate(dataBkg.columns):
    plt.plot(dataBkg.index, dataBkg[column], label=f"PDH{i+1}")
    plt.xticks(rotation=45)
    plt.yscale("log")
plt.ylabel(r"Count Rate [s$^{-1}$]")
plt.legend()
plt.show()

bkgMean = dataBkg.mean(axis=0)
bkgStd = dataBkg.std(axis=0)
print("Bkg Mean")
print(bkgMean)
print("-----------------------------------------")
print("Bkg Standard Deviation")
print(bkgStd)
print("-----------------------------------------")

for i in range(len(listSEPData)):
    listSEPData[i] = listSEPData[i] - bkgMean + 5 * bkgStd
    # neg = listSEPData[i][listSEPData[i] < 0]
    # print(neg.count())
    # listSEPData[i][listSEPData[i] < 0] = 0

# Plot of the raw RADEM count data after removing the background
for dataSEP in listSEPData:
    for i, column in enumerate(dataSEP.columns):
        plt.plot(dataSEP.index, dataSEP[column], label=f"PDH{i+1}")
        plt.xticks(rotation=45)
        plt.yscale("log")
    plt.ylabel(r"Count Rate [s$^{-1}$]")
    plt.legend()
    plt.show()
```

```python
layerWidths = [512, 1024, 4096, 2048, 2048]
 
neuralNetwork = NeuralNetwork(inputDim=nChannels, outputDim=2, layerWidths=layerWidths)
neuralNetwork.load_state_dict(torch.load("../Data/NeuralNetworks/NeuralNetwork_RPLS.pth", map_location=torch.device('cpu')))
device = "cuda" if torch.cuda.is_available() else "cpu"

with open(f"{GetRootPath()}/Data/NeuralNetworks/LogStandardScalers_RPLS.pkl", "rb") as filehandler:
    datasetScalers = pickle.load(filehandler)
featureScaler, targetScaler = datasetScalers

responseFunctionsLoader = ResponseFunctionsLoader()
responseFunctionsLoader.LoadResponseFunctions(option="OldRF")
energies = responseFunctionsLoader.energies
responseFunctions = responseFunctionsLoader.tensorRF
def FluxSpectrumFunc(Energies: torch.tensor, Parameters: torch.tensor):
    return torch.pow(Energies, -Parameters[:, 0].unsqueeze(-1)) * torch.exp(-Energies / Parameters[:, 1].unsqueeze(-1)) 
```

```python
for idx, dataSEP in enumerate(listSEPData):
    initialDate = f"{listSEPDates[idx][0].day}-{listSEPDates[idx][0].month}-{listSEPDates[idx][0].year}"
    finalDate = f"{listSEPDates[idx][1].day}-{listSEPDates[idx][1].month}-{listSEPDates[idx][1].year}"

    countsSEP = torch.tensor(dataSEP.to_numpy(), dtype=torch.float32)
    countsSEP = Dataset(countsSEP, nTargets=0)
    countsSEP.PreprocessData(featureScaler, targetScaler)

    neuralNetwork.to(device)
    neuralNetwork.train()
    nSamples = 100
    index = np.arange(0, len(countsSEP))
    yLabels = [r"Intensity [cm$^{-2}$s$^{-1}$sr$^{-1}$MeV$^{-1}$]", "Spectral Index", "Rolling Energy [MeV]"]
    targetLabels = ["Intensity", "Spectral Index", "Rolling Energy"]

    scaledCounts = countsSEP[:][0].to(device)
    with torch.no_grad():
        predParameters = targetScaler.inverse_transform(neuralNetwork(scaledCounts).cpu())

    # Estimate Intensity with linear regression
    predCounts = FluxParametersToCounts(FluxSpectrumFunc, predParameters, responseFunctions, energies)
    X = predCounts.unsqueeze(-1)
    Y = countsSEP.rawData.unsqueeze(-1)
    predIntensity = torch.bmm(torch.transpose(X, 1, 2), Y).squeeze(-1) / torch.bmm(torch.transpose(X, 1, 2), X).squeeze(-1)
    predParameters = torch.cat((predIntensity, predParameters), dim=-1)

    print("------------------------------------")
    print(f"SEP Event ({listSEPDates[idx][0].day}/{listSEPDates[idx][0].month}/{listSEPDates[idx][0].year} - {listSEPDates[idx][1].day}/{listSEPDates[idx][1].month}/{listSEPDates[idx][1].year})")
    print("------------------------------------")

    for i, yLabel in enumerate(yLabels):
        plt.plot(dataSEP.index, predParameters[:, i])
        plt.yscale("log")
        plt.xticks(rotation=45)
        plt.ylabel(yLabel)
        plt.savefig(f"{GetRootPath()}/Images/PDH_Config1_Reconstruction_RPLS/{targetLabels[i]}_{initialDate}_{finalDate}.png")
        plt.show()

    predCountsSEP = predCounts * predParameters[:, 0].unsqueeze(-1)

    for i in range(predCountsSEP.shape[-1]):
        fig = plt.figure()
        gridSpec = fig.add_gridspec(2, 1, hspace=0.05, wspace=0, height_ratios=[4, 1])
        (ax1, ax2) = gridSpec.subplots(sharex='col')

        ax1.plot(dataSEP.index, predCountsSEP[:, i], label=f"PDH{i+1} Reconstructed")
        ax1.plot(dataSEP.index, countsSEP.rawData[:, i], label=f"PDH{i+1} Data")
        ax1.set_ylabel(r"Count Rate [s$^{-1}$]")
        ax1.set_yscale("log")
        ax1.tick_params(axis="x", which="both", bottom=False, labelbottom=False)
        ax1.legend(loc="lower right")

        residuals = torch.abs(predCountsSEP[:, i] - countsSEP.rawData[:, i]) / countsSEP.rawData[:, i] * 100
        ax2.plot(dataSEP.index, residuals, color="#FF0032")
        ax2.set_ylabel("Relative Error [%]")
        ax2.set_yscale("log")
        ax2.tick_params(axis="x", rotation=45)
        plt.savefig(f"{GetRootPath()}/Images/PDH_Config1_Reconstruction_RPLS/PDH{i+1}_{initialDate}_{finalDate}.png")
        plt.show()

    for i in range(predCountsSEP.shape[-1]):
        plt.plot(predCountsSEP[:, i], countsSEP.rawData[:, i], ".", label=f"PDH{i+1}")
    plt.ylabel(r"Reconstructed Count Rates [s$^{-1}$]")
    plt.xlabel(r"Measured Count Rates [s$^{-1}$]")
    plt.legend()
    plt.show()
```
